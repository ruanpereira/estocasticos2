```python
#importando bibliotecas necessárias
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Lista de tarefas, Q1
```

```python
# parâmetros selecionados para a questão
fs = 1          # Taxa de amostragem (Hz), escolhida arbitrariamente
sigma2 = 1         # Variância do ruído (sigma^2), escolhido arbitrariamente
N_values = [256, 1024, 4096, 16384]

# Valor teórico da DEP
S_w = sigma2 / fs

# Definição do gráfico
fig, axes = plt.subplots(len(N_values), 1, figsize=(10, 10), sharex=True)

print(f"=== RESULTADOS DAS ESTATÍSTICAS DO PERIODOGRAMA ===")
print(f"{'N':<10}{'Variância Estimada':<25}{'Variância Teórica (S_W^2)':<25}")
print("-------------------------------------------------------------")

for i, N in enumerate(N_values):
    # Gerando vetor de N amostras de ruído branco gaussiano w ~ N(0, \sigma^2)
    w = np.random.normal(0, np.sqrt(sigma2), N)
    
    # Fazendo DFT via FFT (transformada discreta de fourier via transformada rápida de fourier, mais eficiente computacionalmente)
    W_fft = np.fft.fft(w)
    
    # Isolando os valores de frequẽncia e calculando o periodograma: P(f) = |FFT|^2 / (N * fs), daí temos também  a definição da escala W/Hz
    freqs = np.fft.fftfreq(N, d=1/fs)
    P_w = (np.abs(W_fft)**2) / (N * fs)
    
    # Fazendo shift para centralizar a freq 0 em f = 0 Hz (-fs/2 a fs/2) (reflexão no eixo negativo e positivo da frequência)
    freqs_shifted = np.fft.fftshift(freqs)
    P_w_shifted = np.fft.fftshift(P_w)
    
    # calculando as variâncias
    var_P = np.var(P_w)
    var_theoretical = S_w ** 2
    
    print(f"{N:<10}{var_P:<25.6e}{var_theoretical:<25.6e}")
    
    # Plot do espectro
    axes[i].plot(freqs_shifted, P_w_shifted, label=f'Periodograma (N={N})', alpha=0.75, color='tab:blue')
    axes[i].axhline(y=S_w, color='r', linestyle='--', linewidth=2, label=f'Teórico ({S_w:.4f})')
    axes[i].set_ylabel("DEP (W/Hz)")
    axes[i].set_title(f"Periodograma para N = {N} amostras (Resolução $\\Delta f = {fs/N:.2f}$ Hz)")
    axes[i].grid(True, linestyle=':', alpha=0.6)
    axes[i].legend(loc='upper right')

axes[-1].set_xlabel("Frequência (Hz)")
plt.tight_layout()
#plt.savefig('grafico_1.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Lista de tarefas, Q2
```

```python
# valores de a decididos arbitrariamente
a_values = [0.2, 0.5, 0.8, 0.9]

# criando a função de filtragem ar
def filtrar_ar1(w, a_val):
    N = len(w)
    y = np.zeros(N)
    y[0] = w[0]
    for n in range(1, N):
        y[n] = a_val * y[n-1] + w[n]
    return y

# Criando a grade de gráficos (4 por 4)
fig, axes = plt.subplots(len(N_values), len(a_values), figsize=(16, 12), sharex=True)
fig.suptitle("Questão 2: DEP do Processo AR via Periodograma vs. Teórico", fontsize=14, y=0.99)

for i, N in enumerate(N_values):
    # Gerando o sinal de ruído branco 
    w = np.random.normal(0, np.sqrt(sigma2), N)
    
    for j, a_val in enumerate(a_values):
        ax = axes[i, j]
        
        # usando a função de filtragem AR
        y = filtrar_ar1(w, a_val)
        
        # calculando o periodograma de y[n]
        Y_fft = np.fft.fft(y)
        freqs = np.fft.fftfreq(N, d=1/fs)
        P_y = (np.abs(Y_fft)**2) / (N * fs)
        
        # Shiftando para centralizar em f = 0 (-fs/2 a fs/2)
        freqs_shifted = np.fft.fftshift(freqs)
        P_y_shifted = np.fft.fftshift(P_y)
        
        # DEP teórica S_Y(f) com base no S_W da questão anterior
        S_Y_theoretical = S_w / (1 - 2*a_val*np.cos(2 * np.pi * freqs_shifted / fs) + a_val**2)
        
        # Plot
        ax.plot(freqs_shifted, P_y_shifted, label='Periodograma', alpha=0.6, color='tab:blue', linewidth=0.8)
        ax.plot(freqs_shifted, S_Y_theoretical, 'r--', linewidth=1.8, label=f'Teórico $S_Y(f)$')
        
        # Títulos e rótulos
        if i == 0:
            ax.set_title(f"a = {a_val}", fontsize=11, fontweight='bold')
        if j == 0:
            ax.set_ylabel(f"N = {N}\nDEP (W/Hz)", fontsize=10)
        
        ax.grid(True, linestyle=':', alpha=0.6)
        if i == 0 and j == 0:
            ax.legend(loc='upper right', fontsize=8)

for j in range(len(a_values)):
    axes[-1, j].set_xlabel("Frequência (Hz)")

plt.tight_layout()
plt.subplots_adjust(top=0.94)
#plt.savefig('grafico_2.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Lista de tarefas, Q3
```

```python
# definindo funções para os estimadores de DEP usados na questão
def periodograma_simples(y, fs):
    N = len(y)
    Y_fft = np.fft.fft(y)
    freqs = np.fft.fftfreq(N, d=1/fs)
    P_y = (np.abs(Y_fft)**2) / (N * fs)
    return np.fft.fftshift(freqs), np.fft.fftshift(P_y)

def metodo_bartlett(y, K, fs):
    N = len(y)
    L = N // K  # Comprimento de cada segmento sem sobreposição
    freqs = np.fft.fftfreq(L, d=1/fs)
    P_acc = np.zeros(L)
    
    for i in range(K):
        segmento = y[i*L : (i+1)*L]
        Y_fft = np.fft.fft(segmento)
        P_acc += (np.abs(Y_fft)**2) / (L * fs)
        
    P_bartlett = P_acc / K
    return np.fft.fftshift(freqs), np.fft.fftshift(P_bartlett)

def metodo_welch_hanning(y, K, fs):
    N = len(y)
    # Para K segmentos com 50% de sobreposição: L = 2N / (K + 1)
    L = int(2 * N / (K + 1))
    hop = L // 2  # Deslocamento entre blocos (50%)
    
    # Janela de Hanning
    m = np.arange(L)
    win = 0.5 - 0.5 * np.cos(2 * np.pi * m / (L - 1))
    U = np.sum(win**2) / L  # Normalização de potência da janela
    
    freqs = np.fft.fftfreq(L, d=1/fs)
    P_acc = np.zeros(L)
    
    for i in range(K):
        inicio = i * hop
        segmento = y[inicio : inicio + L]
        if len(segmento) < L:
            break
        # Aplicação da janela
        segmento_janelado = segmento * win
        Y_fft = np.fft.fft(segmento_janelado)
        P_acc += (np.abs(Y_fft)**2) / (L * fs * U)
        
    P_welch = P_acc / K
    return np.fft.fftshift(freqs), np.fft.fftshift(P_welch)

# definindo os parãmetros de a e N para o filtro AR
a_coef = 0.9        # Coeficiente do AR
N = 4096            # Tamanho fixo do sinal para análise
K_values = [4, 8, 16]  # valores de K, escolhidos como dado na questão

# gerando o sinal com ruído colorido
w = np.random.normal(0, np.sqrt(sigma2), N)
y = filtrar_ar1(w, a_coef)

# usando a função do periodograma simples
freqs_p, P_simples = periodograma_simples(y, fs)

# curva teórica S_Y(f) (da mesma forma que na segunda questão)
S_W = sigma2 / fs
S_Y_theoretical = S_W / (1 - 2*a_coef*np.cos(2 * np.pi * freqs_p / fs) + a_coef**2)

# plot dos gráficos comparativos 
fig, axes = plt.subplots(len(K_values), 1, figsize=(11, 11), sharex=True)
fig.suptitle(f"Questão 3: Comparativo de Métodos de Estimação de DEP ($N = {N}$, $AR$ com $a={a_coef}$)", fontsize=13)

for i, K in enumerate(K_values):
    ax = axes[i]
    
    # Cálculo das estimativas de Bartlett e Welch usando as funções definidas anteriormente
    freqs_b, P_bartlett = metodo_bartlett(y, K, fs)
    freqs_w, P_welch = metodo_welch_hanning(y, K, fs)
    
    # Plotando o periodograma simples no fundo
    ax.plot(freqs_p, P_simples, color='gray', alpha=0.35, linewidth=0.8, label='Periodograma Simples')
    
    # Adicionando Bartlett e Welch
    ax.plot(freqs_b, P_bartlett, color='tab:blue', linewidth=1.8, label=f'Bartlett (K={K})')
    ax.plot(freqs_w, P_welch, color='tab:green', linewidth=2.0, label=f'Welch Hanning (K={K})')
    
    # adicionando a curva teórica
    ax.plot(freqs_p, S_Y_theoretical, 'r--', linewidth=1.8, label=f'Teórico $S_Y(f)$')

    # título, grid, labels, legenda, etc.
    ax.set_title(f"Número de Segmentos K = {K}", fontsize=11, fontweight='bold')
    ax.set_ylabel("DEP (W/Hz)")
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend(loc='upper right', fontsize=9)

axes[-1].set_xlabel("Frequência (Hz)")
plt.tight_layout()
#plt.savefig('grafico_3.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Lista de tarefas, Q4
```

```python
# utilizando a mesma função da terceira questão para welch

def metodo_welch_hanning(y, K, fs):
    N = len(y)
    # para K segmentos com 50% de sobreposição: L = 2N / (K + 1)
    L = int(2 * N / (K + 1))
    hop = L // 2  # deslocamento entre blocos (50%)
    
    # Janela de Hanning, da mesma forma
    m = np.arange(L)
    win = 0.5 - 0.5 * np.cos(2 * np.pi * m / (L - 1))
    U = np.sum(win**2) / L  # normalização de potência da janela

    # separando a frequẽncia na fft
    freqs = np.fft.fftfreq(L, d=1/fs)
    P_acc = np.zeros(L)
    
    for i in range(K):
        inicio = i * hop
        segmento = y[inicio : inicio + L]
        if len(segmento) < L:
            break
        # Aplicação da janela de Hanning
        segmento_janelado = segmento * win
        Y_fft = np.fft.fft(segmento_janelado)
        P_acc += (np.abs(Y_fft)**2) / (L * fs * U)
        
    P_welch = P_acc / K
    return np.fft.fftshift(freqs), np.fft.fftshift(P_welch)

# definindo a função de geração do pulso cosseno levantado
def gerar_pulso_rc(alpha, sps, num_simbolos_span=10):
    # gerando a resposta ao impulso do pulso cosseno levantado.
    t = np.arange(-num_simbolos_span/2, num_simbolos_span/2, 1/sps)
    h = np.zeros_like(t)
    for i, ti in enumerate(t):
        if ti == 0:
            h[i] = 1.0
        elif alpha != 0 and abs(abs(ti) - 1/(2*alpha)) < 1e-8:
            h[i] = (np.pi / 4) * np.sinc(1 / (2 * alpha))
        else:
            denom = 1 - (2 * alpha * ti)**2
            h[i] = np.sinc(ti) * np.cos(np.pi * alpha * ti) / denom
    return h / np.max(h)

# função para adicionar ruido a qualquer sinal
def adicionar_ruido(sinal, snr_db):
    # adicionando AWGN para qualquer SNR em dB
    P_sinal = np.mean(sinal**2)
    snr_linear = 10**(snr_db / 10.0)
    P_ruido = P_sinal / snr_linear
    ruido = np.random.normal(0, np.sqrt(P_ruido), len(sinal))
    return sinal + ruido, P_ruido

# parametros de simulação para BPSK (arbitrario)

Rs = 1000.0             # Taxa de símbolos (1000 Baud)
sps = 8                 # Amostras por símbolo (taxa de sobreamostragem)
fs = Rs * sps           # Taxa de amostragem (8000 Hz)
N_amostras = 10000      # Número total de amostras definido arbitrariamente
num_simbolos = N_amostras // sps

snr_db = 5.0            # SNR = 5 dB
alpha = 0.35            # Fator de roll-off = 0.35
K_segmentos = 16        # Número de segmentos K para a função de Welch

# gerando os simbolos bpsk

# gerando os símbolos BPSK equiprováveis (+1 e -1)
bits = np.random.randint(0, 2, num_simbolos)
simbolos = 2 * bits - 1

# inserindo zeros (Upsampling)
simbolos_up = np.zeros(num_simbolos * sps)
simbolos_up[::sps] = simbolos

# pulso retangular
pulso_ret = np.ones(sps)
sinal_ret = np.convolve(simbolos_up, pulso_ret, mode='same')

# pulso cosseno levantado (roll-off = 0.35)
pulso_rc = gerar_pulso_rc(alpha=alpha, sps=sps)
sinal_rc = np.convolve(simbolos_up, pulso_rc, mode='same')

# adição de ruído para SNR = 5 dB
sinal_ret_ruido, P_ruido_ret = adicionar_ruido(sinal_ret, snr_db)
sinal_rc_ruido, P_ruido_rc = adicionar_ruido(sinal_rc, snr_db)

# estimando DEP com a função welch
# definindo sinais limpos
f_ret, P_ret = metodo_welch_hanning(sinal_ret, K_segmentos, fs)
f_rc, P_rc = metodo_welch_hanning(sinal_rc, K_segmentos, fs)

# definindo sinais com ruído (SNR = 5 dB)
_, P_ret_ruido = metodo_welch_hanning(sinal_ret_ruido, K_segmentos, fs)
_, P_rc_ruido = metodo_welch_hanning(sinal_rc_ruido, K_segmentos, fs)

# curva teórica sinc^2 em dB para comparação
sinc_teorico = (np.sinc(f_ret / Rs))**2
sinc_teorico_dB = 10 * np.log10(sinc_teorico / np.max(sinc_teorico) * np.max(P_ret) + 1e-12)


# plotagem e comparação dos resultados

fig, axes = plt.subplots(2, 1, figsize=(12, 10))

# sinais limpos (dB/Hz)
axes[0].plot(f_ret, 10*np.log10(P_ret + 1e-12), color='tab:blue', label='BPSK Pulso Retangular (Welch K=16)')
axes[0].plot(f_ret, sinc_teorico_dB, 'r--', linewidth=1.5, label=r'Teórico $\propto \text{sinc}^2(f/R_s)$')
axes[0].plot(f_rc, 10*np.log10(P_rc + 1e-12), color='tab:green', linewidth=1.8, label=f'BPSK Cosseno Levantado ($\\alpha={alpha}$)')

axes[0].set_title("Questão 4: DEP Estimada via Método de Welch (Sinais Limpos)", fontsize=12, fontweight='bold')
axes[0].set_ylabel("DEP (dB/Hz)")
axes[0].set_ylim(-60, 10)
axes[0].grid(True, linestyle=':', alpha=0.6)
axes[0].legend(loc='upper right')

# sinais com ruido (SNR = 5 dB) e piso de ruído
axes[1].plot(f_ret, 10*np.log10(P_ret_ruido + 1e-12), color='tab:blue', alpha=0.85, label='BPSK Retangular + Ruído (SNR=5dB)')
axes[1].plot(f_rc, 10*np.log10(P_rc_ruido + 1e-12), color='tab:green', alpha=0.85, label=f'BPSK Cosseno Levantado + Ruído (SNR=5dB)')

# nível do piso de ruído teórico em dB/Hz
piso_ruido_dB = 10 * np.log10(P_ruido_ret / fs)
axes[1].axhline(y=piso_ruido_dB, color='black', linestyle=':', linewidth=2, label=f'Piso de Ruído Teórico ({piso_ruido_dB:.1f} dB/Hz)')

axes[1].set_title("Questão 4: DEP Estimada com Adição de Ruído AWGN (SNR = 5 dB)", fontsize=12, fontweight='bold')
axes[1].set_xlabel("Frequência (Hz)")
axes[1].set_ylabel("DEP (dB/Hz)")
axes[1].set_ylim(-60, 10)
axes[1].grid(True, linestyle=':', alpha=0.6)
axes[1].legend(loc='upper right')

plt.tight_layout()
#plt.savefig('grafico_4.png', dpi=300, bbox_inches='tight')
plt.show()
```
